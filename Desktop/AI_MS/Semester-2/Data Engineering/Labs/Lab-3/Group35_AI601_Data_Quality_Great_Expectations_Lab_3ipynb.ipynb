```python
import great_expectations as gx
import pandas as pd
```

```python
df = pd.read_csv(
    "/content/reddit_healthcare.csv")

df.head()
```

# Using Great Expectations

```python
## Example Setup

context = gx.get_context()
data_source = context.data_sources.add_pandas("pandas")
data_asset = data_source.add_dataframe_asset(name="pd dataframe asset")

batch_definition = data_asset.add_batch_definition_whole_dataframe("batch definition")
batch = batch_definition.get_batch(batch_parameters={"dataframe": df})

#rowlevel
expectation_1 = gx.expectations.ExpectTableRowCountToBeBetween(
    min_value=5, max_value=10000
)

#column level
expectation_2 = gx.expectations.ExpectColumnValuesToBeBetween(
    column="Upvotes", min_value=0
)

#multicolumn level
expectation_3 = gx.expectations.ExpectCompoundColumnsToBeUnique(
    column_list=["Title", "Author"]
)

#Table-Level
expectation_4 = gx.expectations.ExpectTableColumnsToMatchSet(
    column_set=["Title", "Text", "Author", "Date", "Upvotes", "Subreddit"]
)

#distribution
expectation_5 = gx.expectations.ExpectColumnMedianToBeBetween(
    column="Upvotes", min_value=10, max_value=500
)




expectations = [
    expectation_1,
    expectation_2,
    expectation_3,
    expectation_4,
    expectation_5
]

validation_result1 = batch.validate(expectation_1)
validation_result2= batch.validate(expectation_2)
validation_result3 = batch.validate(expectation_3)
validation_result4 = batch.validate(expectation_4)
validation_result5 = batch.validate(expectation_5)

print(validation_result1)
print(validation_result2)
print(validation_result3)
print(validation_result4)
print(validation_result5)

```
